```python
import google.generativeai as genai
```

```python
#waiter
genai.configure(api_key=API_KEY)
```

```python
configuration = {"max_output_tokens": 100,"temperature": 0.4}
model = genai.GenerativeModel("gemini-2.5-flash-lite", generation_config=configuration)
```

```python
# ── Tree of Thought Prompt ─────────────────────────────────────
tot_prompt = """
You are an expert problem solver using the Tree of Thought approach.

For the given question, explore MULTIPLE different reasoning paths 
before arriving at the best answer.

Follow this structure:

BRANCH 1 → First possible way to think about this problem
- Reasoning: ...
- Conclusion: ...
- Confidence: Low / Medium / High

BRANCH 2 → Second possible way to think about this problem
- Reasoning: ...
- Conclusion: ...
- Confidence: Low / Medium / High

BRANCH 3 → Third possible way to think about this problem
- Reasoning: ...
- Conclusion: ...
- Confidence: Low / Medium / High

EVALUATION →
- Which branch has the strongest reasoning?
- Why is it better than the others?

FINAL ANSWER →
Based on the best branch, give your final answer clearly.

Question: {question}
"""

# ── Get Input and Run ToT ──────────────────────────────────────
user_question = input(">Enter the Question: ")

prompt = tot_prompt.format(question=user_question)

response = model.generate_content(prompt)

print(response.text)
```
