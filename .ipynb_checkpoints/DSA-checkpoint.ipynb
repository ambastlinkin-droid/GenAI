```python
def palindromeNumber(n: int) -> bool:
    original=n
    rev=0
    res=0
    while n>0:
        x=n%10
        res=res*10+x
        n//=10
    return original==res

palindromeNumber(121)
    
```

```python
#Check armstrong
def count_the_digits(num: int):
    count=0
    while num>0:
        count+=1
        num//=10
    return count 
```

```python
def check_is_armstrong(num: int):
    power=count_the_digits(num)
    original=num
    sum_of_digits=0
    while num>0:
        rem = num%10
        sum_of_digits+=rem**power
        num//=10
    return original==sum_of_digits
check_is_armstrong(153)
        
```

```python
#divided by 1 or itself
def is_prime(num):
    if num<2:
        return False
    else:
        for i in range(2, int(num**0.5)+1): #till the sqrt of the numbers to find the possible factors
            if num%i==0: #to find the small factors
                return False
        else:
            return True
def print_composite(n):
    for num in range(2,n+1):
        if not is_prime(num):
            print(num)                    
print_composites(10)   
```

```python
def print_composites(n):
    for num in range(2, n+1):
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                print(num, end=" ")
                break  # no need to check further once composite

print_composites(20)
```

```python
#sliding window!
lis = [1,2,1,3,4]
left_ptr = 0
right_ptr = len(lis)-1
isPalindrome = False
while left_ptr<=(len(lis)/2):
    if lis[left_ptr]!=lis[right_ptr]:
        isPalindrome = False
        break
    else:
        isPalindrome = True
    left_ptr+=1
    right_ptr-=1

if isPalindrome:
    print("True")
else:
    print("False")
```

```python
def arrange():
    n = 5 #uncomment this !!
    arr = [0] * n #creating list of n elements
    right_ptr = n-1
    left_ptr = 0
    for i in range(1,n+1):
        if i%2==0:
            arr[right_ptr] = i
            right_ptr-=1
        else:
            arr[left_ptr] = i
            left_ptr+=1
    print(arr)
arrange()
```

```python
def findTriplet(arr, n, x) :
    count=0
    for i in range(n):
        for j in range(i+1,n):
            for k in range(j+1,n):
                if arr[i]+arr[j]+arr[k]==x:
                    count+=1
    return count
t = int(input())

for _ in range(t):
    n = int(input())
    arr = list(map(int, input().split()))
    x = int(input())

    print(findTriplet(arr, n, x))
                
```

```python

def linearSearch(arr, n, val):
    for i in range(len(arr)):
        if arr[i]==val:
            return i
    return -1


if __name__=="__main__":
    n = int(input("Enter length of the array: "))
    arr = list(map(int, input().split()))
    val = int(input("Enter the target"))
    res = linearSearch(arr,n,val)
    print(res)
    
    
```

```python
#find the duplicate element from the array 
arr = [0,7,2,5,4,7,1,3,6]
freq = {}
for i in range(len(arr)):
    freq[arr[i]] = freq.get(arr[i],0)+1
for key,value in freq.items():
    if value>1:
        print(key)
```
