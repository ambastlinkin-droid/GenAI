### Part 1: The "Broken" System (Zero-Shot Prompting)
Here is my initial attempt. I'm using a simple, direct **Zero-Shot prompt** to analyze the reviews.
It works fine for simple cases, but watch what happens when we give it a more nuanced review.

````python
def analyze_review_zero_shot(review_text):
  """Analyzes a review using a simple Zero-Shot prompt."""
  prompt = f"""
  Analyze the sentiment of the following customer review.
  Is it Positive, Negative, or Neutral? Also, provide a brief summary of the reason.

  Review: "{review_text}"

  Sentiment:
  Reason:
  """
  try:
    model = genai.GenerativeModel('gemini-2.5-flash') # Using a standard model
    response = model.generate_content(prompt, generation_config=generation_config)
    # Adding a check to see why a response might be blocked
    if not response.parts:
        return f"The response was blocked. Finish Reason: {response.candidates[0].finish_reason}. Safety Ratings: {response.prompt_feedback.safety_ratings}"
    return response.text
  except Exception as e:
    return f"An error occurred: {e}"
````

````python
# --- Test Case 1: A simple, negative review (This will now work) ---
simple_review = df.loc[3, 'review_text']
print("\n--- ANALYZING A SIMPLE REVIEW ---")
print(f"Review: {simple_review}\n")
print(analyze_review_zero_shot(simple_review))

print("\n" + "="*50 + "\n")

# --- Test Case 2: A complex, mixed-sentiment review (This will now also work) ---
complex_review = df.loc[1, 'review_text']
print("--- ANALYZING A COMPLEX REVIEW ---")
print(f"Review: {complex_review}\n")
print(analyze_review_zero_shot(complex_review))
````

The **Zero-Shot prompt** isn’t reliable enough.

````python
# TODO: Write your Few-Shot prompt here.
# Use 3-4 examples from the dataframe to teach the model the desired JSON format.
# Pay close attention to the mixed-sentiment reviews.

few_shot_prompt = """
<WRITE YOUR FEW-SHOT PROMPT HERE>
"""

def analyze_review_few_shot(review_text):
  """Analyzes a review using the crafted Few-Shot prompt."""
  prompt = few_shot_prompt.format(review_text=review_text)
  try:
    model = genai.GenerativeModel('gemini-2.5-flash')
    response = model.generate_content(prompt)
    # Clean up the output to be valid JSON
    cleaned_response = response.text.strip().replace("```json", "").replace("```", "")
    return json.loads(cleaned_response)
  except Exception as e:
    return {"error": str(e), "raw_response": response.text}


# Let's test it on the most complex review again
complex_review = df.loc[5, 'review_text'] # "Wow! The setup was a breeze... but the battery life is much shorter..."
print(f"--- ANALYZING COMPLEX REVIEW WITH FEW-SHOT ---")
print(f"Review: {complex_review}\n")
result = analyze_review_few_shot(complex_review)
print(json.dumps(result, indent=2))
````

### Part 3: The Advanced Solution (Chain-of-Thought Prompting)
Few-shot is a huge improvement! But for the *most* complex reviews, we can make the model's reasoning even more robust by forcing it to "think step-by-step."

**Your Goal:** Construct a Chain-of-Thought (CoT) prompt. This prompt will explicitly instruct the model on the logical steps it must take *before* giving the final answer. This reduces errors and makes the output more predictable.

**The same aspect list will be used here.**

## Part 4: The Reliability Boost — Self-Consistency

Our **Chain-of-Thought (CoT) prompt** is powerful, but LLMs can be inconsistent.  
Running the exact same prompt multiple times may produce slightly different answers.  

To improve reliability, we use **Self-Consistency**:  
- Run the CoT prompt several times.  
- Collect all the outputs.  
- Choose the most common result as the final answer.

### Comparative Evaluation
Now that we have built three different analysis systems (Zero-Shot, Few-Shot, and CoT), let's run all of them on our full dataset and compare their outputs directly against our "Ground Truth."
This will make it crystal clear why advanced prompting techniques are so powerful.

````python
import time

# Create empty lists to store the results
zero_shot_results = []
few_shot_results = []
cot_results = []

print("Running all analyses with a delay to respect API rate limits...")
for index, row in df.iterrows():
    review_text = row['review_text']
    review_id = row['review_id']

    print(f"  > Analyzing Review #{review_id}...")

    # Run each analysis for the current review
    zero_shot_results.append(analyze_review_zero_shot(review_text))
    few_shot_results.append(analyze_review_few_shot(review_text))
    cot_results.append(analyze_review_cot(review_text))

    time.sleep(1)

# Add the collected results as new columns in the DataFrame
df['zero_shot_result'] = zero_shot_results
df['few_shot_result'] = few_shot_results
df['cot_result'] = cot_results

print("\nAll analyses complete!")
````

````python
from IPython.display import display, HTML

# Helper function to display the results in a nice, readable format
def display_comparison(dataframe):
    for index, row in dataframe.iterrows():
        print("="*80)
        print(f"REVIEW #{row['review_id']}: {row['review_text']}")
        print("-"*80)

        # Ground Truth
        print("✅ GROUND TRUTH:")
        print(json.dumps(row['ground_truth'], indent=2))
        print("\n")

        # Zero-Shot
        print("1️⃣ ZERO-SHOT RESULT:")
        # This result is just text, not JSON
        print(row['zero_shot_result'])
        print("\n")

        # Few-Shot
        print("2️⃣ FEW-SHOT RESULT:")
        # This result is a dictionary
        print(json.dumps(row['few_shot_result'], indent=2))
        print("\n")

        # Chain-of-Thought
        print("3️⃣ CHAIN-OF-THOUGHT RESULT:")
        # This result is also a dictionary
        print(json.dumps(row['cot_result'], indent=2))
        print("\n")

# Display the full comparison
display_comparison(df)
````
