```python
n=int(input())
ans=0
power_of_two=1
while n>0:
    lastdigit=n%10
    ans+=lastdigit*power_of_two
    power_of_two*=2
    n=n//10
print(ans)
```

```python
def main():
    n=int(input())
    ans=0
    rem=0
    power=1
    while n>0:
        rem=n%10
        ans+=rem*power
        
        power*=2
        n//=10
    print(ans)
if __name__=="__main__":
    main()
```
