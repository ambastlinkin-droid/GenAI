```python
#Check armstrong
def count_the_digits(num: int):
    count=0
    while num>0:
        count+=1
        num//=10
    return count 
```

```python
def check_is_armstrong(num: int):
    power=count_the_digits(num)
    original=num
    sum_of_digits=0
    while num>0:
        rem = num%10
        sum_of_digits+=rem**power
        num//=10
    return original==sum_of_digits
check_is_armstrong(153)
        
```

```python
#divided by 1 or itself
def is_prime(num):
    if num<2:
        return False
    else:
        for i in range(2, int(num**0.5)+1): #till the sqrt of the numbers to find the possible factors
            if num%i==0: #to find the small factors
                return False
        else:
            return True
def print_composite(n):
    for num in range(2,n+1):
        if not is_prime(num):
            print(num)                    
  
```

```python
#rotate matrix 90 Degrees anticlockwise!
def rotateMatrix(matrix):
    n = len(matrix)
    m = len(matrix[0])
    #step 1 transpose the matrix
    for i in range(n):
        for j in range(i+1,n):
            matrix[i][j],matrix[j][i]=matrix[j][i],matrix[i][j]
    print("Transpose: ",*matrix)
    #Reverse the each column
    for col in range(m):
        top = 0
        bottom = m-1
        while top<bottom:
            matrix[top][col],matrix[bottom][col] = matrix[bottom][col],matrix[top][col]
            #top<->bottom, bottom<->top
            top+=1 #comes from top to top+1 element
            bottom-=1 #climbs from bottom to top
    return matrix
    
matrix = [
          [2,0,0,1],  
          [0,3,1,0],
          [0,5,2,0], 
          [4,0,0,2]
        ]

#Convert to
"""
#transpose
[2, 0, 0, 4] 
[0, 3, 5, 0] 
[0, 1, 2, 0] 
[1, 0, 0, 2]

#rotated 90Deg anticlock wise
1, 0, 0, 2
0, 1, 2, 0
0, 3, 5, 0
2, 0, 0, 4

"""
print(rotateMatrix(matrix))
```

```python
#print spiral of matrix

def spiralPrint(mat, nRows, mCols):
    if not mat or not mat[0]:
        return []
    else:
        result = []
        top=0
        left=0
        rows = len(mat)
        columns = len(mat[0])
        bottom = rows-1 #rows
        right = columns-1 #cols
        while top<=bottom and left<=right:
            for i in range(left,right+1):
                result.append(mat[top][i])
            top+=1
            for i in range(top,bottom+1):
                result.append(mat[i][right])
            right-=1
            
            if top<=bottom:
                for i in range(right,left-1,-1):
                    result.append(mat[bottom][i])
                bottom-=1

            if left<=right:
                for i in range(bottom,top-1,-1):
                    result.append(mat[i][left])
                left+=1
        print(*result)
```

```python
#compress the string
def getCompressedString(input):
    if input=="":
        return input
    count=1
    result=""
    for i in range(1,len(input)):
        if input[i]==input[i-1]:
            count+=1
        else:
            result+=str(input[i-1])+(str(count) if count>1 else "")
            count=1
    result+=input[-1]+(str(count) if count>1 else "")
    return result
```

```python
#check the max frequency of elements in array
def maxfreq(arr):
    freq = {}
    for elements in arr:
        freq[elements]=freq.get(elements,0)+1
    maximum = max(freq.values())
    for element in arr: #checking the value if it matches with the max value
        if freq[element] == maximum:
            return element

myarr = [1,2,2,1]
print(maxfreq(myarr))
```

```python
#majority element: whose appearance in the arr is more than the len(arr)//2
def majority_element(arr):
    majority = len(arr)//2
    freq={}
    for elements in arr:
        freq[elements]=freq.get(elements,0)+1
    for key,value in freq.items():
        if majority<value:
            return key



myarr = [5,1,5,1,5,5,5,5,5]
print(majority_element(myarr))
```

```python
def helper(test_dict):
    max_unique=0
    max_key=None
    for key,value in test_dict.items():
        unique = len(set(value)) #removing the duplicates to find the unique elements
        if max_unique < unique:
            max_unique = unique
            max_key = key
    return max_key


myDict = {
           "City":[1,2,3,4],
          "Temperature":[25,26,27,28,29],
          "Population":[1000,2000,3000,4000],
          "GDP":[5000,6000,5000,6000]
         }
print(f"{helper(myDict)} has the most unique keys")
```

```python
def find_missing_and_repeating(arr):
    freq={}
    repeated_ele = None
    for elements in arr:
        freq[elements] = freq.get(elements,0)+1
    for key,value in freq.items():
        if value > 1:
            repeated_ele = key
    n = len(arr)
    missing=list(set(range(1, n + 1)) - set(arr)).pop()
    return [missing,repeated_ele]
        
        

myArr = [1,3,5,4,4] #here missing is 2 and repeating is 4 so we have to return that
print(find_missing_and_repeating(myArr))
```
